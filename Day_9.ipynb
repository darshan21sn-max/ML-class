```python
import numpy as np
import pandas as pd
df = pd.read_csv("/content/messy_ecommerce_dataset_120rows.csv")
df
```

```python
df.describe()
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df['Order_ID']=df['Order_ID'].str.strip()
df['Order Date']=np.where(
    df['Order_ID'].str.startswith('ORD-'),
    df['Order_ID'],
    'ORD-' +df['Order_ID']
)
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates() # Drops the entire row having identical values
df=df.drop_duplicates(subset='Order_ID',keep='first')
```

```python
df['Order_ID'].isnull().sum()
```

Customer name

```python
df['Customer_Name']=df['Customer_Name'].str.strip().str.join(" ").str.title()
df['Customer_Name']=df['Customer_Name'].replace(["N/A","Unknown"],np.nan)
```

```python
df['Customer_Name'].isnull().sum()
```

```python
df[df['Customer_Name'].isnull()]
```

Email

```python
df['Email']=df['Email'].str.strip().str.lower()
valid_email=(df['Email'].str.contains('@',na=False)) & (df['Email'].str.contains('.com',na=False,regex=False))
df.loc[~valid_email,'Email']=np.nan
```

```python
df['Email'].isnull().sum()
```

```python
df[df['Email'].isnull()]
```

Phone Number

```python
digits=df['Phone'].str.replace(r'\D','',regex=True)
has_country_code = (
    (digits.str.len() == 11) &
    digits.str.startswith('1', na=False)
)
digits=digits.where(~has_country_code,digits.str[1:])

valid_phone=(
    (digits.str.len()==10) &
    (digits != '0000000000') &
    digits.notna()
)

df['Phone']=np.where(
    valid_phone,
    digits.str[:3] + '-' + digits.str[3:6] + '-' + digits[6:] ,
    np.nan)

```

```python
df['Phone'].isnull().sum()
```

```python
df[df['Phone'].isnull()]
```

Product Catogery

```python
df['Product_Category'].unique()
```

```python
df['Product_Category']=df['Product_Category'].str.strip().str.lower()

df['Product_Category'] = df['Product_Category'].replace(
    ['n/a', 'na', 'unknown', 'none', ''],
    np.nan
)

category_mapping = {
    'electronics': 'Electronics',
    'home and kitchen': 'Home & Kitchen',
    'home & kitchen': 'Home & Kitchen',
    'beauty & health': 'Beauty & Health',
    'clothing': 'Clothing',
    'fashion': 'Clothing',
    'books': 'Books',
    'groceries': 'Groceries',
    'beauty': 'Beauty',
    'toys': 'Toys',
    'sports': 'Sports',
    'apparel': 'Apparel'
}

df['Product_Category']=df['Product_Category'].replace(category_mapping)

print(df['Product_Category'].value_counts(dropna=False))
```

```python
df[df['Product_Category'].isnull()]
```

City

```python
df['City']=df['City'].str.strip().str.title().replace({
    'NY':"New York",
    "LA" :"Los Angeles",
    "CHI" : "Chicago",
    "SF" : "San Francisco"
})
```

```python
df[df['City'].isnull()]
```

Membership Status

```python
df['Membership_Status']=df['Membership_Status'].str.strip().str.title().replace({'Unknown':np.nan}).fillna('Unassigned')
```

```python
df[df['Membership_Status'].isnull()]
```

Payment Method

```python
df['Payment_Method']=df['Payment_Method'].str.strip().str.lower().replace({
    'cc':'credit card',
    'dc':'debit card'
}).fillna('Unassigned')
```

```python
df['Payment_Method'].isnull().sum()
```

Age

```python
df['Age']=df['Age'].str.strip().replace(r'\D','',regex=True)
df['Age']=pd.to_numeric(df['Age'],errors='coerce')

df.loc[df['Age']<=0, 'Age']=np.nan
df.loc[df['Age']>120, 'Age']=np.nan
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median()).astype(int)
```

```python
df['Age'].isnull().sum()
```

Quantity

```python
df['Quantity']=df['Quantity'].str.strip().replace({
    'two':2,
    'three':3
})


df['Quantity']=pd.to_numeric(df['Quantity'],errors='coerce')

df.loc[df['Quantity']<=0,'Quantity']=np.nan
df.loc[df['Quantity']>500,'Quantity']=np.nan

df['Quantity']=df['Quantity'].astype('Int64')
```

```python
df['Quantity'].unique()
```

```python
df['Quantity'].isnull().sum()
```

```python
df[df['Quantity'].isnull()]
```

```python
df['Unit_Price']=(
    df['Unit_Price']
    .str.replace("$","",regex=False)
    .str.replace('USD','',regex=False)
)

df['Unit_Price']=df['Unit_Price'].str.replace(",",".")

df['Unit_Price']=pd.to_numeric(
    df['Unit_Price'],
    errors='coerce'
)
```

```python
df.loc[df['Unit_Price']<0,"Unit_Price"]=np.nan
df.loc[df['Unit_Price']==999999.99,"Unit_Price"]=np.nan
```

```python
df['Unit_Price'].isnull().sum()
```

```python
df[df['Quantity'].isnull()]
```

Discount

```python
df['Discount'] = df['Discount'].astype('string').str.strip()

df['Discount'] = df['Discount'].str.replace('%', '', regex=False)

df['Discount'] = pd.to_numeric(df['Discount'], errors='coerce')

df.loc[df['Discount'] > 1, 'Discount'] = df['Discount'] / 100

df.loc[
    (df['Discount'] < 0) | (df['Discount'] > 1),
    'Discount'
] = np.nan

df['Discount'] = df['Discount'].fillna(0.00)
```

```python
df['Discount'].isnull().sum()
```

Date

```python
df['Order_Date'] = pd.to_datetime(
    df['Order_Date'],
    errors='coerce',
    format='mixed'
)

valid_dates = (
    (df['Order_Date'] >= '2023-01-01') &
    (df['Order_Date'] <= '2023-12-31')
)

df.loc[~valid_dates, 'Order_Date'] = pd.NaT

df['Order_Date'] = df['Order_Date'].dt.strftime('%Y-%m-%d')
```

```python
df['Order_Date'].isnull().sum()
```

```python
df[df['Order_Date'].isnull()]
```

```python
df['Total_Sales']=df['Quantity']* df['Unit_Price']* (1-df['Discount'])
df['Total_Sales'].fillna(0)
```

```python
df.isnull().sum()
```

```python
df = df.dropna(
    subset=['Order_ID', 'Unit_Price'])
```

```python
df.isnull().sum()
```

```python
df.to_csv('cleaned_ecommerce_dataset.csv', index=False)
```

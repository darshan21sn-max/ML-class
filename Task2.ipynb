```python
def clean_customer_name(name):

    # Check missing values
    if pd.isna(name):
        return np.nan

    # Convert to string
    name = str(name)

    # Remove leading and trailing spaces
    name = name.strip()

    # Replace multiple spaces with single space
    name = re.sub(r"\s+", " ", name)

    # Replace placeholder values
    placeholders = [
        "n/a", "na", "null",
        "none", "unknown", "?"
    ]

    if name.lower() in placeholders:
        return np.nan

    # Convert into Title Case
    name = name.title()

    return name


# Apply cleaning function
df["Customer_Name"] = df["Customer_Name"].apply(
    clean_customer_name
)

print("\nCleaned Customer Names:")
print(df["Customer_Name"].head(20))
```

```python
# Email validation function

def clean_email(email):

    # Check missing values
    if pd.isna(email):
        return np.nan

    # Convert to string
    email = str(email).strip().lower()

    # Replace placeholders
    placeholders = [
        "n/a", "na", "null",
        "none", "unknown", "?"
    ]

    if email in placeholders or email == "":
        return np.nan

    return email


# Apply cleaning
df["Email"] = df["Email"].apply(clean_email)


# Email validation function

def validate_email(email):

    if pd.isna(email):
        return "Missing"

    # Email validation pattern
    pattern = r"^[a-z0-9._%+-]+@[a-z0-9.-]+\.[a-z]{2,}$"

    if re.match(pattern, email):
        return "Valid"

    else:
        return "Invalid"


# Create validation column
df["Email_Status"] = df["Email"].apply(
    validate_email
)

# Replace invalid email addresses with NaN
df.loc[
    df["Email_Status"] == "Invalid",
    "Email"
] = np.nan


print("\nCleaned Emails:")
print(
    df[["Email", "Email_Status"]].head(20)
)

print("\nEmail Validation Summary:")
print(df["Email_Status"].value_counts())
```

```python
# Phone number cleaning function

def clean_phone(phone):

    # Check missing values
    if pd.isna(phone):
        return np.nan

    # Convert into string
    phone = str(phone).strip()

    # Replace placeholders
    placeholders = [
        "n/a", "na", "null",
        "none", "unknown", "invalid_phone",
        "?", ""
    ]

    if phone.lower() in placeholders:
        return np.nan

    # Remove trailing .0 from numeric-looking values
    phone = re.sub(r"\.0$", "", phone)

    # Remove all punctuation and formatting characters
    digits = re.sub(r"\D", "", phone)

    # Remove leading +1 country code
    # Only when number contains 11 digits
    # and begins with 1
    if len(digits) == 11 and digits.startswith("1"):
        digits = digits[1:]

    # Check valid 10-digit number
    if len(digits) != 10:
        return np.nan

    # Reject all-zero phone numbers
    if digits == "0000000000":
        return np.nan

    # Format XXX-XXX-XXXX
    formatted_phone = (
        digits[:3] + "-" +
        digits[3:6] + "-" +
        digits[6:]
    )

    return formatted_phone


# Apply phone cleaning
df["Phone"] = df["Phone"].apply(clean_phone)

print("\nCleaned Phone Numbers:")
print(df["Phone"].head(20))

print("\nInvalid or Missing Phone Count:")
print(df["Phone"].isna().sum())
```

```python
def clean_text(value):

    if pd.isna(value):
        return np.nan

    # Convert to string
    value = str(value)

    # Remove leading and trailing spaces
    value = value.strip()

    # Replace multiple spaces with one space
    value = re.sub(r"\s+", " ", value)

    # Replace placeholders with NaN
    placeholders = [
        "N/A",
        "NA",
        "NULL",
        "NONE",
        "UNKNOWN",
        "",
        "?"
    ]

    if value.upper() in placeholders:
        return np.nan

    return value
```

```python
# Extract numeric value from Age
def clean_age(value):

    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    # Extract the numeric portion
    match = re.search(r"-?\d+(?:\.\d+)?", value)

    if match:
        return float(match.group())

    return np.nan


df["Age"] = df["Age"].apply(clean_age)

# Convert to numeric
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

# Remove impossible ages
df.loc[(df["Age"] <= 0) | (df["Age"] > 120), "Age"] = np.nan

# Calculate median age
median_age = df["Age"].median()

print("Median Age:", median_age)

# Impute missing ages
df["Age"] = df["Age"].fillna(median_age)

# Convert to integer
df["Age"] = df["Age"].round().astype("Int64")

print(df["Age"].describe())
print("Missing Ages:", df["Age"].isnull().sum())
```

```python
# Written number mapping
number_words = {
    "zero": 0,
    "one": 1,
    "two": 2,
    "three": 3,
    "four": 4,
    "five": 5,
    "six": 6,
    "seven": 7,
    "eight": 8,
    "nine": 9,
    "ten": 10
}


def clean_quantity(value):

    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    # Convert written numbers
    if value in number_words:
        return number_words[value]

    # Extract numeric value
    match = re.search(r"-?\d+", value)

    if match:
        return int(match.group())

    return np.nan


df["Quantity"] = df["Quantity"].apply(clean_quantity)

# Convert into numeric
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")

# Flag invalid quantities
df["Quantity_Invalid"] = df["Quantity"].isna() | (df["Quantity"] <= 0)

# Flag extreme bulk quantities
df["Quantity_Outlier"] = df["Quantity"] > 100

# Remove invalid quantities
df.loc[df["Quantity_Invalid"], "Quantity"] = np.nan

# Keep extreme quantities but flag them for review
print("Invalid Quantity Rows:", df["Quantity_Invalid"].sum())
print("Extreme Quantity Rows:", df["Quantity_Outlier"].sum())

print(df[["Quantity", "Quantity_Invalid", "Quantity_Outlier"]].head(15))
```

```python
def clean_unit_price(value):

    if pd.isna(value):
        return np.nan

    value = str(value).strip().upper()

    # Remove currency codes and symbols
    value = re.sub(r"USD", "", value)
    value = re.sub(r"[$€£₹]", "", value)

    # Remove spaces
    value = value.replace(" ", "")

    # Handle European and US number formats
    if "," in value and "." in value:

        # Example: 1,250.75
        if value.rfind(".") > value.rfind(","):
            value = value.replace(",", "")

        # Example: 1.250,75
        else:
            value = value.replace(".", "")
            value = value.replace(",", ".")

    elif "," in value:

        # Example: 125,50 -> 125.50
        if len(value.split(",")[-1]) == 2:
            value = value.replace(",", ".")

        # Example: 1,250 -> 1250
        else:
            value = value.replace(",", "")

    return pd.to_numeric(value, errors="coerce")


df["Unit_Price"] = df["Unit_Price"].apply(clean_unit_price)

# Flag negative prices
df["Price_Invalid"] = df["Unit_Price"].isna() | (df["Unit_Price"] < 0)

# Convert negative prices to NaN
df.loc[df["Unit_Price"] < 0, "Unit_Price"] = np.nan

# Flag unusually high prices
df["Price_Outlier"] = df["Unit_Price"] > 10000

print("Invalid Prices:", df["Price_Invalid"].sum())
print("Extreme Prices:", df["Price_Outlier"].sum())

print(df[["Unit_Price", "Price_Invalid", "Price_Outlier"]].head(15))
```

```python
def clean_discount(value):

    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    if value == "":
        return np.nan

    # Check whether the original value contains %
    is_percentage = "%" in value

    # Remove percentage symbol
    value = value.replace("%", "").strip()

    # Convert to numeric
    number = pd.to_numeric(value, errors="coerce")

    if pd.isna(number):
        return np.nan

    # Convert percentage into decimal
    if is_percentage or number > 1:
        number = number / 100

    return number


df["Discount"] = df["Discount"].apply(clean_discount)

# Flag impossible discounts
df["Discount_Invalid"] = (
    df["Discount"].isna() |
    (df["Discount"] < 0) |
    (df["Discount"] > 1)
)

# Replace impossible discounts with NaN
df.loc[
    (df["Discount"] < 0) | (df["Discount"] > 1),
    "Discount"
] = np.nan

# Fill missing discounts with zero
df["Discount"] = df["Discount"].fillna(0.00)

print("Invalid Discount Rows:", df["Discount_Invalid"].sum())

print(df["Discount"].value_counts(dropna=False))
```

```python
from datetime import datetime

def clean_order_date(value):

    if pd.isna(value):
        return pd.NaT

    value = str(value).strip()

    # Empty and placeholder text
    if value.lower() in ["", "n/a", "na", "null", "none", "unknown", "?"]:
        return pd.NaT

    # Supported date formats
    date_formats = [
        "%Y-%m-%d",   # 2024-01-15
        "%m/%d/%Y",   # 01/15/2024
        "%d-%m-%Y",   # 15-01-2024
        "%Y/%m/%d",   # 2024/01/15
        "%b %d, %Y",  # Jan 15, 2024
        "%B %d, %Y"   # January 15, 2024
    ]

    for fmt in date_formats:
        try:
            return pd.Timestamp(datetime.strptime(value, fmt))
        except ValueError:
            continue

    return pd.NaT


df["Order_Date"] = df["Order_Date"].apply(clean_order_date)

print(df[["Order_Date"]].head(15))
```

```python
# Define placeholder dates
placeholder_dates = [
    pd.Timestamp("1900-01-01"),
    pd.Timestamp("2099-12-31")
]

# Create validation flags
df["Date_Invalid"] = df["Order_Date"].isna()

df["Date_Placeholder"] = df["Order_Date"].isin(placeholder_dates)

# Identify dates after today's date
today = pd.Timestamp.today().normalize()

df["Date_Future"] = df["Order_Date"] > today

# Nullify placeholder dates
df.loc[df["Date_Placeholder"], "Order_Date"] = pd.NaT

# Nullify unrealistic future dates
df.loc[df["Date_Future"], "Order_Date"] = pd.NaT

# Update invalid status after all checks
df["Date_Invalid"] = df["Order_Date"].isna()

print("Placeholder Dates:", df["Date_Placeholder"].sum())
print("Future Dates:", df["Date_Future"].sum())
print("Invalid or Missing Dates:", df["Date_Invalid"].sum())
```

PHASE 7

```python
import pandas as pd
import numpy as np
import re

# ==========================================
# STEP 1: USE IN-MEMORY DATASET
# ==========================================
# Instead of loading from a non-existent CSV, we use the 'df' already in memory
rows_before = len(df)
columns_before = len(df.columns)

print("Before Cleaning:", df.shape)


# ==========================================
# STEP 2: CURRENCY CLEANING FUNCTION
# ==========================================

def clean_currency(value):

    if pd.isna(value):
        return np.nan

    value = str(value).strip().upper()

    if value in ["", "N/A", "NA", "NULL", "NONE", "UNKNOWN", "?"]:
        return np.nan

    # Remove currency codes
    value = re.sub(r"\bUSD\b", "", value)
    value = re.sub(r"\bINR\b", "", value)
    value = re.sub(r"\bEUR\b", "", value)

    # Remove currency symbols
    value = re.sub(r"[$€£₹]", "", value)

    # Remove spaces
    value = value.replace(" ", "")

    # Handle different decimal formats
    if "," in value and "." in value:

        if value.rfind(".") > value.rfind(","):
            # Example: 1,250.50
            value = value.replace(",", "")
        else:
            # Example: 1.250,50
            value = value.replace(".", "")
            value = value.replace(",", ".")

    elif "," in value:

        decimal_part = value.split(",")[-1]

        if len(decimal_part) in [1, 2]:
            # Example: 125,50
            value = value.replace(",", ".")
        else:
            # Example: 1,250
            value = value.replace(",", "")

    return pd.to_numeric(value, errors="coerce")


# ==========================================
# STEP 3: CLEAN TOTAL_SALES AND UNIT_PRICE
# ==========================================

# Preserve original Total_Sales for auditing
if "Total_Sales" in df.columns:
    df["Total_Sales_Original" ] = df["Total_Sales"].copy()

    # Clean original currency formatting
    df["Total_Sales_Original"] = (
        df["Total_Sales_Original"].apply(clean_currency)
    )
else:
    df["Total_Sales_Original"] = np.nan

# Clean Unit_Price
df["Unit_Price"] = df["Unit_Price"].apply(clean_currency)


# ==========================================
# STEP 4: VALIDATE QUANTITY
# ==========================================

df["Quantity"] = pd.to_numeric(
    df["Quantity"], errors="coerce"
)

# Quantity must be a positive whole number
valid_quantity = (
    df["Quantity"].notna()
    & (df["Quantity"] > 0)
    & (df["Quantity"] % 1 == 0)
)

df.loc[~valid_quantity, "Quantity"] = np.nan


# ==========================================
# STEP 5: CLEAN DISCOUNT
# ==========================================

def clean_discount(value):

    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    if value == "":
        return np.nan

    has_percent = "%" in value

    value = value.replace("%", "")

    number = pd.to_numeric(value, errors="coerce")

    if pd.isna(number):
        return np.nan

    # Convert percentage to decimal
    if has_percent or number > 1:
        number = number / 100

    # Validate discount range
    if number < 0 or number > 1:
        return np.nan

    return float(number)


df["Discount"] = df["Discount"].apply(clean_discount)

# Fill missing or invalid discounts with zero
df["Discount"] = df["Discount"].fillna(0.00)


# ==========================================
# STEP 6: HANDLE CRITICAL MISSING VALUES
# ==========================================

# Clean Order_ID
df["Order_ID"] = df["Order_ID"].astype("string").str.strip()

invalid_ids = df["Order_ID"].str.lower().isin(
    ["", "n/a", "na", "null", "none", "unknown", "?"]
)

df.loc[invalid_ids, "Order_ID"] = pd.NA

# Identify invalid transaction rows
invalid_order_id = df["Order_ID"].isna()

invalid_price = (
    df["Unit_Price"].isna()
    | (df["Unit_Price"] < 0)
)

invalid_quantity = (
    df["Quantity"].isna()
    | (df["Quantity"] <= 0)
)

# Count rows for reporting
print("Missing Order_ID:", invalid_order_id.sum())
print("Missing/Invalid Unit_Price:", invalid_price.sum())
print("Missing/Invalid Quantity:", invalid_quantity.sum())

# Remove rows with invalid critical fields
drop_rows = (
    invalid_order_id
    | invalid_price
    | invalid_quantity
)

rows_removed = int(drop_rows.sum())

df = df.loc[~drop_rows].copy()


# ==========================================
# STEP 7: FLAG EXTREME VALUES
# ==========================================

# Keep suspicious values for review
df["Quantity_Outlier"] = df["Quantity"] > 100
df["Unit_Price_Outlier"] = df["Unit_Price"] > 10000


# ==========================================
# STEP 8: RECALCULATE TOTAL_SALES
# ==========================================

df["Total_Sales"] = (
    df["Quantity"]
    * df["Unit_Price"]
    * (1 - df["Discount"])
)

df["Total_Sales"] = df["Total_Sales"].round(2)


# ==========================================
# STEP 9: HANDLE NON-CRITICAL MISSING VALUES
# ==========================================

# Fill missing categorical attributes
categorical_columns = df.select_dtypes(
    include=["object", "string", "category"]
).columns

for column in categorical_columns:

    if column not in ["Order_ID", "Order_Date"]:
         df[column] = df[column].replace(
            r"^\s*$", np.nan, regex=True
         )

         df[column] = df[column].fillna("Unassigned")


# ==========================================
# STEP 10: FINAL VALIDATION
# ==========================================

rows_after = len(df)
columns_after = len(df.columns)

print("\n========== FINAL DATASET ==========")

print("Rows Before:", rows_before)
print("Rows After:", rows_after)
print("Rows Removed:", rows_removed)

print("Columns Before:", columns_before)
print("Columns After:", columns_after)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nData Types:")
print(df.dtypes)

print("\nTotal Sales Summary:")
print(df["Total_Sales"].describe())


# ==========================================
# STEP 11: SAVE FINAL DATASET
# ==========================================

df.to_csv(
    "cleaned_ecommerce_dataset.csv",
    index=False
)

print("\nPhase 7 Completed Successfully!")
print("Saved: cleaned_ecommerce_dataset.csv")
```

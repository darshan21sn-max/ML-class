```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
import pandas as pd
from sklearn.datasets import load_iris
df = pd.read_csv("Iris.csv")
df
iris = load_iris()

#box plot
sns.boxplot(x=df['SepalLengthCm'])
plt.show()

#line chart
sns.lineplot(x=df['SepalLengthCm'], y=df['SepalWidthCm'])
plt.show()

#scatter plot
sns.scatterplot(x=df['SepalLengthCm'], y=df['SepalWidthCm'])
plt.show()

#bar plot
sns.barplot(x=df['SepalLengthCm'], y=df['SepalWidthCm'])
plt.show()

#histogram
sns.histplot(x=df['SepalLengthCm'])
plt.show()
```

```python
plt.plot(df['SepalLengthCm'], df['SepalWidthCm'])
plt.xlabel('SepalLengthCm')
plt.ylabel('SepalWidthCm')
plt.title('Scatter plot')
plt.show()
```

```python
sns.barplot(x=df['SepalLengthCm'], y=df['SepalWidthCm'], color='red')
plt.xlabel('SepalLengthCm')
plt.ylabel('SepalWidthCm')
plt.title('Bar plot')
plt.show()
```

```python
sns.lineplot(x=df['SepalLengthCm'], y=df['SepalWidthCm'], color='green')
plt.xlabel('Sepal length')
plt.ylabel('Sepal width')
plt.title('Line plot')
plt.show()
```

```python
sns.pairplot(df, vars=['SepalLengthCm', 'SepalWidthCm'], hue='Species', palette='Set1')
plt.title('Pair plot', y=1.02)
plt.show()
```

```python
plt.pie(df['Species'].value_counts(), labels=df['Species'].unique(), autopct='4.5', colors=sns.color_palette('Set3'))
plt.title('Distribution of Iris Species')
plt.show()
```

```python
sns.scatterplot(data=df, x='SepalLengthCm', y='SepalWidthCm', sizes=(20, 200), alpha=0.7, palette='Set1')
plt.title('Iris Bubble Plot')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.show()
```

```python
sns.countplot(x=df['Species'],color ="red")
plt.show()
```

```python
sns.violinplot(x=df['Species'], y=df['SepalLengthCm'], color= 'yellow')
plt.show()
```
